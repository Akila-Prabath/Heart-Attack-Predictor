```python
import os
import joblib
import pandas as pd
import numpy as np
from xgboost import XGBClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report, confusion_matrix
```

```python
DATA_PATH = "../../data/raw/heart_attack_prediction.csv"
MODELS_DIR = "../../models"
os.makedirs(MODELS_DIR, exist_ok=True)
```

```python
df = pd.read_csv(DATA_PATH)
```

```python
categorical_cols = df.select_dtypes(include=['object']).columns
le = LabelEncoder()
for col in categorical_cols:
    df[col] = le.fit_transform(df[col].astype(str))
```

```python
X = df.drop('heart_attack', axis=1)
y = df['heart_attack']
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)
```

```python
print("Training data shape:", X_train.shape)
print("Test data shape:", X_test.shape)
```

```python
model = XGBClassifier(eval_metric='logloss', random_state=42)
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
unique_vals = np.unique(y_train)
avg_method = 'binary' if len(unique_vals) == 2 and set(unique_vals) <= set([0,1]) else 'weighted'
```

```python
acc = accuracy_score(y_test, y_pred)
prec = precision_score(y_test, y_pred, average=avg_method, zero_division=0)
rec = recall_score(y_test, y_pred, average=avg_method, zero_division=0)
f1 = f1_score(y_test, y_pred, average=avg_method, zero_division=0)
```

```python
print("\nModel evaluation (XGBoost):")
print(f" - Accuracy:  {acc:.4f}")
print(f" - Precision: {prec:.4f}")
print(f" - Recall:    {rec:.4f}")
print(f" - F1-score:  {f1:.4f}")
```

```python
print("\nClassification report:")
print(classification_report(y_test, y_pred, zero_division=0))
```

```python
print("Confusion matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
joblib.dump(model, os.path.join(MODELS_DIR, 'xgboost_model.pkl'))
joblib.dump(scaler, os.path.join(MODELS_DIR, 'scaler.pkl'))
print("\nSaved XGBoost model and scaler to:", MODELS_DIR)
```
